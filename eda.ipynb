### Importing the dataset CIFAR-10

```python
from src.dataset import get_eda_dataset

dataset = get_eda_dataset()

dataset
```

---

### Which are the classes in the dataset?

```python
class_names = dataset.classes

for i in class_names:
    print(i)
```

---

### How is the data balanced?

```python
import pandas as pd

labels = dataset.targets
label_names = [class_names[i] for i in labels]

class_counts = pd.Series(label_names).value_counts().sort_index()
class_counts
```

### Analysis: Each class contains 5000 samples, indicating a perfectly balanced dataset.

---

### Plotting examples of the pictures in the dataset

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

for i in range(5):
    img, label = dataset[i]

    plt.subplot(1, 5, i + 1)
    plt.imshow(img.permute(1, 2, 0))
    plt.title(class_names[label])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

### Analysis: The images are low resolution (32×32), resulting in limited detail, which may restrict performance on more complex tasks.
